## Linear Regression

 Lignear Regression via Gradient Descent

$$
\hat{y} = w_1x + w_0
$$

$$
L_B(\mathbf{w})
=
\frac{1}{2|B|}
\sum_{(x_i,y_i)\in B}
(y_i-\hat{y}_i)^2
$$

```python
# Dataset
x = [1.0, 2.0, 3.0, 4.0]
y = [2.0, 3.0, 5.0, 8.0]

# Hyperparameters
learning_rate = 0.2
num_epochs = 100
batch_size = 4

# Initial parameters
w0 = 1.0
w1 = 2.0

num_samples = len(x)

for epoch in range(num_epochs):
    for start_idx in range(0, num_samples, batch_size):

        x_batch = x[start_idx:start_idx + batch_size]
        y_batch = y[start_idx:start_idx + batch_size]

        # 1) Forward
        y_hat = []

        for i in range(len(x_batch)):
            y_hat.append(w1 * x_batch[i] + w0)

        # 2) Loss
        loss = 0.0

        for i in range(len(x_batch)):
            error = y_hat[i] - y_batch[i]
            loss = loss + 0.5 * error ** 2

        loss = loss / len(x_batch)  # loss =  1/(2 * B) * (y - y_hat)^2

        ##########################
        # 3) Compute gradients for the batch

        # gard_w0 = dL/dw0 = 1/B * sum(y_hat - y)
        # grad_w1 = dL/dw1 = 1/B * sum((y_hat - y) * x)

        grad_w0 = 0.0
        grad_w1 = 0.0

        for i in range(len(x_batch)):
            error = y_hat[i] - y_batch[i]
            grad_w0 += error
            grad_w1 += error * x_batch[i]

        grad_w0 = grad_w0 / len(x_batch)
        grad_w1 = grad_w1 / len(x_batch)

        ##########################

        # 4) Update
        w0 = w0 - learning_rate * grad_w0
        w1 = w1 - learning_rate * grad_w1

print("w0 =", w0)
print("w1 =", w1)
```

## PyTorch

This time, we implement Gradient Descent using PyTorch.

`nn.Linear(1, 1)` represents the following model.
$$
\hat{y} = w_1x + w_0
$$

```python
import torch
import torch.nn as nn

# Dataset
x = torch.tensor([[1.0], [2.0], [3.0], [4.0]])
y = torch.tensor([[2.0], [3.0], [5.0], [8.0]])

# Hyperparameters
learning_rate = 0.1
num_epochs = 100
batch_size = 1

num_samples = len(x)

# Model: y_hat = w1*x + w0
model = nn.Linear(1, 1)

# Initial parameters: w1 = 2, w0 = 1
with torch.no_grad():
    model.weight.fill_(2.0)
    model.bias.fill_(1.0)

optimizer = torch.optim.SGD(model.parameters(), lr=learning_rate)  # Gradient Descent

for epoch in range(num_epochs):
    for start_idx in range(0, num_samples, batch_size):

        x_batch = x[start_idx:start_idx + batch_size]
        y_batch = y[start_idx:start_idx + batch_size]

        # 1) Forward
        y_hat = model(x_batch)

        # 2) Loss
        # L = 1/(2B) * sum((y - y_hat)^2)
        loss = 0.5 * torch.mean((y_hat - y_batch) ** 2)

        # 3) Gradient
        optimizer.zero_grad()
        loss.backward()

        # 4) Update
        optimizer.step()

print("w0 =", model.bias[0])
print("w1 =", model.weight[0, 0])
```

```python
import torch
import torch.nn as nn

# Dataset
x = torch.tensor([[1.0], [2.0], [3.0], [4.0]])
y = torch.tensor([[2.0], [3.0], [5.0], [8.0]])

# Hyperparameters
learning_rate = 0.2
num_epochs = 100
batch_size = 1

num_samples = len(x)

# Model: y_hat = w1*x + w0
model = nn.Linear(1, 1)

# Initial parameters: w1 = 2, w0 = 1
with torch.no_grad():
    model.weight.fill_(2.0)
    model.bias.fill_(1.0)

# Optimizer
optimizer = torch.optim.Adam(
    model.parameters(),
    lr=learning_rate
)

# Training
for epoch in range(num_epochs):
    for start_idx in range(0, num_samples, batch_size):

        x_batch = x[start_idx:start_idx + batch_size]
        y_batch = y[start_idx:start_idx + batch_size]

        # 1) Forward
        y_hat = model(x_batch)

        # 2) Loss
        # L = 1/(2B) * sum((y - y_hat)^2)
        loss = 0.5 * torch.mean((y_hat - y_batch) ** 2)

        # 3) Gradient
        optimizer.zero_grad()
        loss.backward()

        # 4) Update
        optimizer.step()

# Final parameters
print("w0 =", model.bias[0].item())
print("w1 =", model.weight[0, 0].item())

# 전체 데이터에 대한 최종 성능 확인
with torch.no_grad():
    y_pred = model(x)

    # MSE
    mse = torch.mean((y_pred - y) ** 2)

    # 학습에서 사용한 Loss = 0.5 * MSE
    final_loss = 0.5 * mse

print("\nPrediction")
print(y_pred)

print("\nMSE =", mse.item())
print("Loss =", final_loss.item())
```
